```python
filePath = "D:\\Study\\GitRepository\\Pyspark-projects\\Basic-projects\\ReadCSVFile\\SampleSuperstore.csv"
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
        .appName("ReadCSV") \
        .getOrCreate()
```

```python
dfSuperStore = spark.read.csv(filePath,header=True, inferSchema=True)

dfSuperStore.show()
```

### Print schema

```python
dfSuperStore.printSchema()
```

### Select specific columns: Select City, Sales, and Profit

```python
from pyspark.sql.functions import col

df_specificColumns = dfSuperStore.select(col("City").alias("City") \
                              ,col("Sales").alias("Sales") \
                              ,col("Profit").alias("Profit"))

df_specificColumns.limit(10).show()
```

## Filter rows by Region

```python
dfSuperStore.filter(dfSuperStore["Region"] == "West").show(5)
```

## Rename a column

```python
df_specificColumns.withColumnRenamed("Profit", "NetProfit").show(3)
```

## Sort by Sales descending:

```python
dfSuperStore.orderBy("Sales", descending = True).show(3)
```

## Cast data type

```python
dfSuperStore.withColumn("discount",dfSuperStore["discount"].cast("float")).show(10)
```

## Add a constant column

```python
from pyspark.sql.functions import lit

dfSuperStore.withColumn("Company", lit("SuperStore")).show(5)
```

## Check nulls

```python
dfSuperStore.filter(dfSuperStore["Sub-Category"].isNull()).show(10)
```

## Group by Region and Sum Sales. Apply Round and do alias 

```python
from pyspark.sql.functions import aggregate, sum, round

dfSuperStore.groupBy("Region").agg(round(sum("Sales"),2).alias("Sales")).show(10)
```

## Count orders per Segment

```python
dfSuperStore.groupBy("Segment").count().show()
```
