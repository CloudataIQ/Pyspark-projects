## The steps to itendify how many words in a text.

 In our case there is a paragraph and few texts with in the paragraph and converting.

We are using the below technices to use the program to run.

1. SparkSession - Spark session is to run the program using Spark environment.



```python
import os

PYTHON_PATH = r"c:\ProgramData\anaconda3\python.exe"

os.environ["PYSPARK_PYTHON"] = PYTHON_PATH
os.environ["PYSPARK_DRIVER_PYTHON"] = PYTHON_PATH
```

```python
paragraph = """
Artificial intelligence, or AI, is when machines are designed to think or act like humans. It helps computers solve problems, understand language, and even recognize images or sounds. For example, when you talk to a voice assistant on your phone and it replies with useful information, that’s AI in action. It’s not magic—it’s the result of computers learning from data.

AI works by finding patterns in large amounts of information. Just like people learn from experience, machines can improve by analyzing what they see over time. This allows them to make better choices without needing new instructions every time. That’s how your music app seems to know what songs you might like or how maps suggest the best routes.

You can find AI in many areas of daily life. It helps recommend movies, sorts your emails, and even powers smart cameras. In hospitals, AI can help doctors spot problems in X-rays. On the road, it helps cars drive themselves more safely. Even farming uses AI to check crop health and improve harvests.

The main idea behind AI is to assist people, not replace them. It takes over boring or complicated jobs so people can focus on creative or important work. While it still has limits, AI is getting better all the time. And although it might seem like something from the future, it’s already changing the way we live and work today.
"""
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
        .appName("WordCount") \
        .getOrCreate()

text_rdd = spark.sparkContext.parallelize(paragraph.strip().split('\n'))

word_rdd = text_rdd.flatMap(lambda line: line.split())

import re
clean_words_rdd = word_rdd.map(lambda word: re.sub(r'\W+', '', word.lower())).filter(lambda w: w)

word_pairs = clean_words_rdd.map(lambda word: (word, 1))

word_counts = word_pairs.reduceByKey(lambda a, b: a + b)

results = word_counts.collect()

for word, count in results:
    print (f"{word} : {count}")
```
